# Fantasy Points Index — Form Tracker (Players Currently in UCL 2026/27)

The same four position-specific equations from `fantasy_points_index.ipynb`, applied across **three timeframes** side by side — filtered by **player identity**, not by which club a given stat line was recorded at.

**The filter rule:** a player is included if their *current* (2026/27) club is in this season's Champions League — even if their strong previous-season or previous-UCL numbers were racked up at a different club (a transfer). So Grimaldo (now at Atlético, in this year's UCL) keeps his excellent Bayer Leverkusen numbers from 2025/26, even though Leverkusen themselves didn't qualify this year. Same for Anthony Gordon (now Barcelona, previously Newcastle) and Rodri (now Barcelona, previously Man City).

1. **Previous domestic season (2025/26), all leagues combined** — who was in top form last year
2. **Previous UCL (2025/26)** — who delivered specifically in the Champions League
3. **Current domestic season so far (2026/27)** — who's in form *right now* (small sample — 2-5 games per player)

Your squad is highlighted in red throughout. Equations (unchanged):

```
FwdIndex = 2×Apps + 4×Goals + 3×Assists + ⌊Tackles/3⌋
MidIndex = 2×Apps + 5×Goals + 3×Assists + ⌊Tackles/3⌋ + 1×CleanSheets
DefIndex = 2×Apps + 6×Goals + 3×Assists + ⌊Tackles/3⌋ + 4×CleanSheets
GkIndex  = 2×Apps + 6×Goals + 3×Assists + 4×CleanSheets + ⌊Saves/3⌋
```

**How the "currently at a UCL club" list was built:** every player who appears in any 2026/27 domestic top-50 leaderboard (goals/assists/tackles/clean sheets/saves/rating/interceptions) for a club in this year's UCL, plus your 15 squad members added explicitly (some, like Rodri and Nuno Mendes, haven't cracked a current-season leaderboard yet but are confirmed still at qualifying clubs).

**Caveat on coverage:** this pulls from a ~165-player pool, and defenders are under-represented in it (they rarely top goals/assists lists) — so the Defenders sections below are noticeably thinner than Forwards/Midfielders, especially for previous UCL. Team names shown in the charts' data are the club where that particular stat was earned, which may be a former club, not necessarily their 2026/27 team.

Same caveats as before: `2×Apps` approximates appearance+60-min points, tackles proxy for recoveries, no POTM/penalties/cards data. The "combined domestic" leaderboards merge Premier League, Bundesliga, LaLiga and Ligue 1 into one ranking, which mixes leagues of different strength — a simplification, not a precise comparison.

````python
import matplotlib.pyplot as plt
import math

plt.rcParams['figure.dpi'] = 110
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False

# Squad as of Matchday 2 transfers
squad = {"Harry Kane", "Mbeumo", "Fabio Silva", "Fábio Silva", "Raphinha", "Bruno Fernandes", "Vitinha", "Angelo Stiller", "Rodri",
         "Gabriel Magalhães", "Alejandro Grimaldo", "Grimaldo", "Nico Schlotterbeck", "Nuno Mendes", "Maguire",
         "Manuel Neuer", "Chevalier"}

def fwd_index(apps, goals, assists, tackles):
    return 2*apps + 4*goals + 3*assists + math.floor(tackles/3)

def mid_index(apps, goals, assists, tackles, clean_sheets):
    return 2*apps + 5*goals + 3*assists + math.floor(tackles/3) + 1*clean_sheets

def def_index(apps, goals, assists, tackles, clean_sheets):
    return 2*apps + 6*goals + 3*assists + math.floor(tackles/3) + 4*clean_sheets

def gk_index(apps, goals, assists, clean_sheets, saves):
    return 2*apps + 6*goals + 3*assists + 4*clean_sheets + math.floor(saves/3)

def plot_index(title, rows):
    rows = rows[:15]
    names_ = [r[0] for r in rows]
    vals = [r[1] for r in rows]
    bar_colors = ["#e63946" if n in squad else "#8d99ae" for n in names_]
    fig, ax = plt.subplots(figsize=(8, 6))
    ax.barh(names_[::-1], vals[::-1], color=bar_colors[::-1])
    ax.set_title(title)
    ax.set_xlabel("Fantasy Points Index (estimated)")
    plt.tight_layout()
    plt.show()
````
