# Top 15 Players by Fantasy Stat, per Position — 2025/26 UCL

League-wide leaderboards (not just your squad) for the stats that drive UEFA Champions League Fantasy points, split by position — Forwards, Midfielders, Defenders, Goalkeepers.

Data: Sofascore's 2025/26 UCL top-players statistics. **Your squad players are highlighted in red** wherever they appear.

**Data limits:** the underlying leaderboard only surfaces goalkeepers for clean sheets (outfield players' clean sheet tallies aren't separately ranked by Sofascore), and tackles/interceptions are shown as the closest available proxy for ball-recovery activity (Fantasy's exact "ball recovered" stat isn't a published leaderboard category).

```python
import matplotlib.pyplot as plt
import pandas as pd

plt.rcParams['figure.dpi'] = 110
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False

squad = {"Isak", "Mbeumo", "Harry Kane", "Raphinha", "Mainoo", "Rodri", "Vitinha", "Anthony Gordon",
         "Gabriel Magalhães", "Alejandro Grimaldo", "Maguire", "Nuno Mendes", "Davies", "Manuel Neuer", "Lucas Chevalier"}

def plot_top15(title, rows, unit=""):
    rows = rows[:15]
    names_, vals = zip(*rows)
    bar_colors = ["#e63946" if n in squad else "#8d99ae" for n in names_]
    fig, ax = plt.subplots(figsize=(8, 6))
    ax.barh(names_[::-1], vals[::-1], color=bar_colors[::-1])
    ax.set_title(title)
    ax.set_xlabel(unit)
    plt.tight_layout()
    plt.show()
```
