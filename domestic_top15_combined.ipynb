# Top 15 Players by Stat, per Position — 2025/26 Domestic, UCL 2026/27 Clubs Only

Premier League, Bundesliga, LaLiga and Ligue 1 merged into single top-15 leaderboards per position and stat — **filtered to only players whose club is in this season's (2026/27) Champions League league phase.** Your squad players are highlighted in red wherever they appear.

This filter removes players from clubs not in this year's UCL — e.g. **Bayer Leverkusen missed out on qualifying for 2026/27**, so Alejandro Grimaldo's 2025/26 Leverkusen numbers are excluded here (he now plays for Atlético Madrid, who are in the competition, but last season's stats were racked up at a non-qualifying club).

**Caveat:** combining leagues of different competitiveness into one ranking is a simplification — treat this as a compiled view, not a precise cross-league ranking.

```python
import matplotlib.pyplot as plt

plt.rcParams['figure.dpi'] = 110
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False

squad = {"Isak", "Mbeumo", "Kane", "Raphinha", "Mainoo", "Rodri", "Vitinha", "Gordon",
         "Gabriel Magalhães", "Grimaldo", "Maguire", "Nuno Mendes", "Davies", "Neuer", "Chevalier"}

def plot_top15(title, rows, unit=""):
    rows = rows[:15]
    names_, vals = zip(*rows)
    bar_colors = ["#e63946" if n in squad else "#8d99ae" for n in names_]
    fig, ax = plt.subplots(figsize=(8, max(4, len(rows)*0.4)))
    ax.barh(names_[::-1], vals[::-1], color=bar_colors[::-1])
    ax.set_title(title)
    ax.set_xlabel(unit)
    plt.tight_layout()
    plt.show()
```
