# Fantasy Points Index — Top 15 by Position (2025/26 UCL)

Four separate composite equations — one per position — built directly from UCL Fantasy's own scoring rules (see `notes.md`), applied to real 2025/26 Champions League stats. Each position uses only the variables that actually earn it points. Your squad is highlighted in red.

## The four equations

**Forwards** (goals worth 4pts, no clean sheet bonus):
```
FwdIndex = 2×Apps + 4×Goals + 3×Assists + ⌊Tackles/3⌋
```

**Midfielders** (goals worth 5pts, clean sheet worth 1pt):
```
MidIndex = 2×Apps + 5×Goals + 3×Assists + ⌊Tackles/3⌋ + 1×CleanSheets
```

**Defenders** (goals worth 6pts, clean sheet worth 4pts):
```
DefIndex = 2×Apps + 6×Goals + 3×Assists + ⌊Tackles/3⌋ + 4×CleanSheets
```

**Goalkeepers** (goals worth 6pts, clean sheet worth 4pts, plus saves):
```
GkIndex = 2×Apps + 6×Goals + 3×Assists + 4×CleanSheets + ⌊Saves/3⌋
```

`2×Apps` stands in for the Appearance point (1) + the 60-minutes point (1) combined — an approximation since exact minutes-per-appearance aren't in this dataset (see caveats). `⌊Tackles/3⌋` stands in for Fantasy's "every 3 balls recovered = +1pt" rule, using tackles as the closest available proxy for recoveries — goalkeepers don't get a recoveries term since they rarely tackle, but do get `⌊Saves/3⌋` for the "every 3 saves = +1pt" rule instead.

**Not included** (no reliable data source): Player of the Match, penalties won/missed/conceded, cards, goals-outside-the-box bonus, goals-conceded penalty. So every index below *understates* true Fantasy points somewhat, but should rank players in roughly the right order for how much position-relevant output they produced.

**Caveats:**
- The appearance/60-min estimate assumes most matches played were starts with 60+ minutes — inflates players who mostly came on as late subs.
- Tackles is a proxy for "ball recoveries" (not a published Sofascore stat), so recovery points are almost certainly undercounted vs. the real Fantasy figure.
- Only players who appeared in at least one of Sofascore's UCL top-50 leaderboards (goals/assists/tackles/clean sheets/saves) are included — a very quiet player who racked up appearance points alone wouldn't show up here.

````python
import matplotlib.pyplot as plt
import math

plt.rcParams['figure.dpi'] = 110
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False

squad = {"Harry Kane", "Anthony Gordon", "Alejandro Grimaldo", "Vitinha", "Nuno Mendes", "Manuel Neuer",
         "Isak", "Mbeumo", "Raphinha", "Mainoo", "Rodri", "Gabriel Magalhães", "Maguire", "Davies", "Lucas Chevalier"}

# Four separate equations, one per position — each using only the variables that actually score points for that position.

def fwd_index(apps, goals, assists, tackles):
    return 2*apps + 4*goals + 3*assists + math.floor(tackles/3)

def mid_index(apps, goals, assists, tackles, clean_sheets):
    return 2*apps + 5*goals + 3*assists + math.floor(tackles/3) + 1*clean_sheets

def def_index(apps, goals, assists, tackles, clean_sheets):
    return 2*apps + 6*goals + 3*assists + math.floor(tackles/3) + 4*clean_sheets

def gk_index(apps, goals, assists, clean_sheets, saves):
    return 2*apps + 6*goals + 3*assists + 4*clean_sheets + math.floor(saves/3)

def plot_index(title, rows):
    rows = rows[:15]
    names_ = [r[0] for r in rows]
    vals = [r[1] for r in rows]
    bar_colors = ["#e63946" if n in squad else "#8d99ae" for n in names_]
    fig, ax = plt.subplots(figsize=(8, 6))
    ax.barh(names_[::-1], vals[::-1], color=bar_colors[::-1])
    ax.set_title(title)
    ax.set_xlabel("Fantasy Points Index (estimated)")
    plt.tight_layout()
    plt.show()
````
